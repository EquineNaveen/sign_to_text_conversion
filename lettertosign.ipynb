```python
import os
import cv2
```

```python
def display_sign_images(word, base_path):
    images = []
    
    for char in word.lower():
        char_path = os.path.join(base_path, char, f"{char}.jpg")  # Assuming images are named as 'a.jpg', 'b.jpg', etc.
        
        if os.path.exists(char_path):
            img = cv2.imread(char_path)
            images.append(img)
        else:
            print(f"Image not found for character: {char}")
    
    for img in images:
        if img is not None:
            cv2.imshow("Sign Language Image", img)
            cv2.waitKey(1000)  # Display each image for 1 second
            cv2.destroyAllWindows()
            
```

```python
if __name__ == "__main__":
    base_directory = "D:/temp/test/dataset"  # Update with the correct path
    input_word = input("Enter a word: ")
    display_sign_images(input_word, base_directory)
    
```
